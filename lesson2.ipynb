```python
# Pandas Data Frame
# we can grab data using reaad_csv or read_excel etc
sp500 = pd.read_csv('StockData/sp500.csv')
sp500['Date'] # we may need to convert strings to actual dates
sp500['Date'] = pd.to_datetime( sp500['Date'] , format='%m/%d/%y') # a date-time picture
```

```python
# we may wish to store our prepared data
sp500.to_excel('StockData/sp500_clean.xlsx')
```

```python
# make the index based on the 'Date' column
sp500.set_index( 'Date', inplace=True ) # we specify the column as the index (column gets removed)
```

```python
sp500
```

```python
# slicing
sp500.loc['20130930':'20131007'] # start:stop INCLUSIVE
```

```python
# we may also use the underlying ordinal index with iloc
sp500.iloc[15:18] # we're back to start:stop-before
```

```python
# we may also choose which columns we wish to see
sp500['Volume'] # for a single column, just use []
sp500[['Volume', 'High']] # for more than one column use [[]] ( the inner [] is a list )
```

```python
# how could we show just October 2013 only Open and Close
sp500.loc['20131001':'20131031'][['Open', 'Close']]
```

```python
# we may apply operations to our data
sp500['Volume'].sum()
sp500['Volume'].mean()
sp500.info()
```

```python
sp500.describe()
```

```python
# sort, filter, moving average
sp500.sort_values(['Volume', 'Close'], ascending=False) # sort by one or more columns
sp500 # we may choose to include inplace=True
```

```python
# Filtering
sp500[ sp500['Close'] > sp500['Open'] ] 
sp500[ sp500['High'] > sp500['Close'] ] 
# mabye we only need above a threshold
# use & for logical-and, also pipe-character for logical-OR
sp500[ (sp500['Volume'] > 1.30e+09) & (sp500['Volume'] < 3.50e+09) ]
```

```python
# we may need to generate new columns containung values derived from our data
sp500['logClose'] = np.log( sp500['Close'] )
sp500
```

```python
# Moving Averages
sp500['ma_1'] = sp500['Close'].shift(1)
sp500['ma_5'] = sp500['Close'].rolling( window=5, min_periods=5 ).mean()
sp500['ma_20'] = sp500['Close'].rolling( window=20, min_periods=20 ).mean()
sp500.head(40)
```
