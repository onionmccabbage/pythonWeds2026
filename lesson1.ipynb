```python
# the List collection
tickers = ['aapl', 'sp500', 'msft', 'nflx']
print( type(tickers) )
tickers[:3] # default start is 0
# we can work with a list
tickers[2] = 'FB'
tickers.insert(2, 'BA') # position to insert, value
tickers.append('IBM')
tickers.pop()
tickers
```

```python
# more than one list...
others = ['GOOG', 'HA']
tickers.append(others)
tickers.pop()
# we can unpck the collection...
tickers.extend(others)
tickers
```

```python
# the tuple collection (immutable)
details = ('Floella', 'Benjamin', 'Dame')
details
```

```python
# Looping over members of a collection
for i in details:
    print(i)
```

```python
# there is a non-ordinal collection called a dictionary
# key:value pairs
person = {'fn':'Ethel', 'sn':'Skronk', 'age':98, 'acc':['current', 'saving', 'ISA']}
type(person)
person['age'] +=1
person
```

```python
# range for series of values
r = range(-100, 101, 20) # start, stop-before, step=1
for i in r:
    print(i**2) # + - * / also ** raise to power
```

```python
# Conditinal Logic
initCap = 300
interest = 0.34
target = 400
# > < != == and or
if (initCap*interest) > target:
    print('retire')
elif initCap*interest > 101:
    print('yay!')
else:
    print('not yet')
```

```python
# Defining Functions
def capReturn(c, p):
    '''We often write a documentation string (docstring)
    Declare what this functin will do
    Take a value and a percentage and return the calculation'''
    result = (c*p) + c
    return result

# use the funtion
q = capReturn(300, 0.35)
r = capReturn(100, 0.35)
s = capReturn(300**4, 0.35)

print(q, r, s)

caps = [100, 200, 300 ,400]
inte = [0.35, 0.000001, 0.42]
for c in caps:
    r = c*inte[0] + c
    s = c*inte[1] + c
    t = c*inte[2] + c
    print(f'{r:0.2f}, {s:0.2f}, {t:0.2f}')
```
