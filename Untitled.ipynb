```python
# Pandas Data Frame
# we can grab data using reaad_csv or read_excel etc
sp500 = pd.read_csv('StockData/sp500.csv')
sp500['Date'] # we may need to convert strings to actual dates
sp500['Date'] = pd.to_datetime( sp500['Date'] , format='%m/%d/%y') # a date-time picture
```

```python
# we may wish to store our prepared data
sp500.to_excel('StockData/sp500_clean.xlsx')
```

```python
# slicing
sp500.loc['20130930':'20131007'] # start:stop INCLUSIVE
```

```python
# we may also use the underlying ordinal index with iloc
sp500.iloc[15:18] # we're back to start:stop-before
```

```python
# we may also choose which columns we wish to see
sp500['Volume'] # for a single column, just use []
sp500[['Volume', 'High']] # for more than one column use [[]] ( the inner [] is a list )
```

```python
# how could we show just October 2013 only Open and Close
sp500.loc['20131001':'20131031'][['Open', 'Close']]
```
