```python
# Data Types
# a string is an ordinal collection of characters
hello = 'Welcome to Python for Finance'
hello
# we use slicing to access members of a collection
hello[11:17] # start:stop-before
```

```python
# the List collection
tickers = ['aapl', 'sp500', 'msft', 'nflx']
print( type(tickers) )
tickers[:3] # default start is 0
# we can work with a list
tickers[2] = 'FB'
tickers.insert(2, 'BA') # position to insert, value
tickers.append('IBM')
tickers.pop()
tickers
```

```python
# more than one list...
others = ['GOOG', 'HA']
tickers.append(others)
tickers.pop()
# we can unpck the collection...
tickers.extend(others)
tickers
```
